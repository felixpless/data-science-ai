### Why Reusable Functions Matter

In professional analytics projects, the same calculations are often required across different datasets, departments, or reporting periods.

Instead of repeating calculations, we encapsulate them in reusable functions.

**Example:** A conversion rate measures the proportion of visitors who complete a purchase.

Conversion Rate = Purchases / Visitors

When expressed as a percentage, multiply the result by 100.



### Exercise 02.09.01.01 — Build a Reusable Conversion Rate Function

Create a function called `conversion_rate()` that:

- Accepts `purchases` and `visitors` as integers.
- Accepts an optional `as_percentage` argument, defaulting to `False`.
- Returns a floating-point value.
- Raises `ValueError` if visitors are zero or negative.
- Raises `ValueError` if purchases are negative or exceed visitors.
- Returns either a decimal conversion rate or a percentage.
- Includes a descriptive docstring.

**Expected results:**

| Input | Expected output |
|---|---:|
| 120 purchases, 1,000 visitors | 0.12 |
| 120 purchases, 1,000 visitors, percentage mode | 12.0 |
| 150 purchases, 1,000 visitors, percentage mode | 15.0 |


```python

def conversion_rate(
    purchases: int,
    visitors: int,
    as_percentage: bool = False
) -> float:
    """
    Calculate the conversion rate from purchases and visitors.

    Args:
        purchases: Number of completed purchases.
        visitors: Total number of visitors.
        as_percentage: Return a percentage if True.

    Returns:
        Conversion rate as a decimal or percentage.

    Raises:
        ValueError: If visitors are not positive or purchases
                    are outside the valid range.
    """
    if visitors <= 0:
        raise ValueError("Visitors must be positive")

    if purchases < 0 or purchases > visitors:
        raise ValueError(
            "Purchases must be between 0 and visitors"
        )

    rate = purchases / visitors

    if as_percentage:
        return rate * 100

    return rate


# Test cases
print(conversion_rate(120, 1000))
print(conversion_rate(120, 1000, as_percentage=True))
print(conversion_rate(150, 1000, as_percentage=True))
```


### Key Concepts

**Type hints**

`purchases: int` indicates that purchases should be an integer.

`-> float` indicates that the function is expected to return a float.

Python type hints do not automatically enforce types at runtime.

**Default parameters**

`as_percentage: bool = False` makes percentage conversion optional.

**Input validation**

Validation prevents invalid inputs from producing misleading business metrics.

**Logical operators**

In ordinary Python conditions, use `and`, `or`, and `not`.

In Pandas, element-wise boolean operations commonly use `&`, `|`, and `~`, with conditions enclosed in parentheses.

**Docstrings**

Docstrings document a function's purpose, parameters, return values, and possible exceptions.

### Business Scenario

A logistics company receives shipment revenue data from different systems.

Some revenue values are stored as numbers, others as strings, and some are missing or invalid.

Our objective is to convert valid revenue values into floating-point numbers while preserving invalid or unknown values as missing.

**Important:** Never invent financial values to replace missing revenue.



### Exercise 02.09.02.01 — Parse Revenue Values

Create a function called `parse_revenue()`.

Requirements:

- Accept strings, integers, floats, and `None`.
- Convert valid values into floats.
- Return `None` for values that cannot be converted.
- Reject negative revenue.
- Reject non-finite numbers such as NaN and infinity.
- Handle `ValueError` and `TypeError`.

Use `try` and `except` to prevent invalid values from crashing the function.


```python

import math


def parse_revenue(
    value: str | int | float | None
) -> float | None:
    """
    Convert valid revenue values into floats.

    Invalid, negative, or non-finite values return None.
    """
    try:
        revenue = float(value)

    except (ValueError, TypeError):
        return None

    if not math.isfinite(revenue):
        return None

    if revenue < 0:
        return None

    return revenue


# Test cases
test_values = [
    "1000",
    250,
    125.50,
    "invalid",
    None,
    -100,
    float("nan"),
    float("inf")
]

for value in test_values:
    print(value, "->", parse_revenue(value))
```
